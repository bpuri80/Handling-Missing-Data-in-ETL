| Method                | Description                                                                           | Appropriate Scenario                                                                                                                               |
| --------------------- | ------------------------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Listwise Deletion** | Removes entire rows (records) that contain one or more missing values.                | Appropriate when the dataset has very few missing values and removing those rows will not significantly reduce the dataset size or introduce bias. |
| **Column Deletion**   | Removes entire columns (attributes) that contain a high percentage of missing values. | Appropriate when a column has a large amount of missing data and is not critical for analysis or decision-making.                                  |

Q4. Why is median imputation preferred over mean imputation for skewed data such as income?

ans4: Median imputation is preferred over mean imputation for skewed data (such as income) because the median is not affected by extreme values or outliers, whereas the mean can be heavily influenced by very high or very low values. As a result, median imputation provides a more representative estimate of the central tendency for skewed distributions and helps preserve the dataset's overall pattern.**

Q5. What is forward fill and in what type of dataset is it most useful?

ans 5 : Forward Fill is a data imputation technique where a missing value is replaced with the most recent non-missing value that appears before it.

Q6. Why should flagging missing values be done before imputation in an ETL workflow?

ans6: Flagging missing values before imputation is important because it preserves information about which values were originally missing. After imputation, it becomes impossible to distinguish between original and filled values. This helps maintain data transparency, supports better analysis, and can improve the performance of machine learning models by allowing them to recognize patterns related to missing data.

Q7. Consider a scenario where income is missing for many customers.
How can this missingness itself provide business insights?

ans 7: Missing income values can itself provide valuable business insights because the missingness may not be random. For example, customers who choose not to disclose their income may belong to specific demographic groups, have privacy concerns, or represent certain income segments. Analyzing these patterns can help businesses improve customer profiling, personalize marketing strategies, assess risk more accurately, and identify gaps in data collection processes.
